```python
from pathlib import Path
import py7zr
import os

os.makedirs("attachments", exist_ok=True)
sourceFolder = "data/DATA - have i been pwned - 14.08.2024"
py7zr.SevenZipFile(
    "data/DATA - have i been pwned - 07.05.2025/HIBP 8 digit wordlist heat.7z"
).extractall(path=sourceFolder)


def convertFile(digits):
    Path(f"data/pins_{digits}.csv").write_text(
        "pin,n\n"
        + Path(f"{sourceFolder}/HIBP {digits} digit wordlist heat.txt")
        .read_text()
        .replace(" ", "")
        .replace(":", ",")
    )


convertFile(3)

convertFile(4)

convertFile(5)

convertFile(6)

convertFile(8)
```
